```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB, BernoulliNB
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, classification_report
from sklearn.svm import SVC
```

# Load Data

```python
df = pd.read_csv("data/SMSSpamCollection.csv")
```

```python
df.head()
```

```python
x = df["message"]
y = df["label"]
```

```python
df["message"].isnull().sum()
```

```python
x_train, x_test, y_train,y_test = train_test_split(x, y, random_state=42, test_size=0.15)
```

```python
x_train.shape, x_test.shape
```

# Naive Bayes

```python
vect = TfidfVectorizer(lowercase=True, stop_words="english", ngram_range=(1,2), min_df=2, sublinear_tf=True)
```

```python
x_train_vect = vect.fit_transform(x_train)
x_test_vect = vect.transform(x_test)
```

```python
model_mul = MultinomialNB()

model_mul.fit(x_train_vect, y_train)
y_pred = model_mul.predict(x_test_vect)
```

```python
print(classification_report(y_test, y_pred))
```

```python
accuracy_score(y_test, y_pred)
```

# SVM 

```python
model_svc = SVC(C=1, kernel="linear")
```

```python
model_svc.fit(x_train_vect, y_train)
y_pred = model_svc.predict(x_test_vect)
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(model_svc, x_train_vect, y_train, cv=5)
print(scores.mean())
```

```python
accuracy_score(y_test, y_pred)
```

```python
print(classification_report(y_test, y_pred))
```

# Logistic Regression

```python
lr_model = LogisticRegression()

lr_model.fit(x_train_vect, y_train)

lr_pred = lr_model.predict(x_test_vect)

print("Logistic Regression Accuracy:", accuracy_score(y_test, lr_pred),"\n")
print(classification_report(y_test, lr_pred))
```
